### Question 1 – Prime Time ⏰

Complete the implementation of the function `prime_time_logins`, which takes in a DataFrame like `login` and outputs a DataFrame indexed by `'Login Id'`, counting the number of prime-time logins for each user – that is, the number of logins that were between 4PM (inclusive) and 8PM (exclusive) for each user. The DataFrame should have just one column, named `'Time'`.

For example, if a user logs in at 5PM on Day 1, at 1PM on Day 2, at 6PM on Day 2, and at 7PM on Day 2, then their total number of prime-time logins is 3. Note that the values in your returned DataFrame should only include counts, not timestamp objects.

***Note***: You do not need to use Python's `datetime` module – instead, use the built-in `pandas` methods for working with times that we introduced in our lectures (though you may need to do a bit more research to fully answer the question).

````python
def prime_time_logins(login):
    """
    Takes in a DataFrame `login` containing 'Login Id' and 'Time' columns.
    Returns a DataFrame indexed by 'Login Id' with a column 'Time' representing
    the count of logins between 4PM (inclusive) and 8PM (exclusive).
    """
    # 1. Ensure the 'Time' column is in datetime format
    # This allows us to access specific time components like hour, minute, etc.
    login['Time'] = pd.to_datetime(login['Time'])
    
    # 2. Extract the hour component
    # The .dt accessor gives us access to datetime properties of the Series
    hours = login['Time'].dt.hour
    
    # 3. Create a boolean mask for Prime Time
    # 4 PM is 16:00, 8 PM is 20:00.
    # We want 16 <= hour < 20.
    is_prime_time = (hours >= 16) & (hours < 20)
    
    # 4. Group by User and Count
    # We group the boolean Series 'is_prime_time' using the 'Login Id' from the original dataframe.
    # Summing a boolean series treats True as 1 and False as 0, effectively counting the matches.
    # This approach ensures users with 0 prime-time logins are still included in the result (showing 0).
    result_series = is_prime_time.groupby(login['Login Id']).sum()
    
    # 5. Format the Output
    # Convert the Series to a DataFrame and rename the column to 'Time' as requested.
    result_df = result_series.to_frame(name='Time')
    
    return result_df.astype(int)
````

### Question 2 – Return Users 🔁

As a site owner, you would like to find your most enthusiastic users – the ones who return to your site most frequently. You've noticed that there are users who have several logins and users who logged in only once. You are interested in finding the number of logins *per day* for each user – that is, **their total number of logins divided by the total number of days they've been on your site**.

Complete the implementation of the function `count_frequency`, which takes in a DataFrame like `login` and outputs a Series containing the number of logins per day for each user. Your Series should have `'Login Id'`s in its index, and the frequencies as its values. The order of users in the index is arbitrary.

To do this, you should assume today is January 31st, 2024 and use 11:59:00PM as the current time. The first login date of a user is the first day of their membership on the site, and you may assume they are still a member today. For simplicity, you only need to count full days that a user has been a member till the end of today. For example, if a user's first login was 12 days and 5 hours ago, you can say that they have been a user for 12 days. 

***Hint***: Can you write a custom aggregator that allows you to do this with just one call of the `groupby` method?

````python
def count_frequency(login):
    """
    Calculates the number of logins per day for each user.
    
    Args:
        login (DataFrame): DataFrame with 'Login Id' and 'Time' columns.
        
    Returns:
        Series: Index is 'Login Id', values are logins per day (float).
    """
    # 1. Ensure Time is in datetime format
    login['Time'] = pd.to_datetime(login['Time'])
    
    # 2. Define the "Current Time" as specified in the prompt
    current_time = pd.Timestamp('2024-01-31 23:59:00')

    # 3. Define a custom aggregator function
    # This function will receive the Series of 'Time' values for *each* user individually
    def calculate_user_freq(user_times):
        # Total number of logins is just the number of timestamps
        total_logins = len(user_times)
        
        # The user's start date is their earliest login time
        first_login = user_times.min()
        
        # Duration is the difference between NOW and the first login
        # This creates a Timedelta object
        duration = current_time - first_login
        
        # Extract full days (as requested by the prompt)
        days_member = duration.days
        
        # Avoid division by zero if the user joined less than 24 hours ago
        # (Though in this specific scenario, we assume historical data)
        if days_member == 0:
            return total_logins # Or appropriate fallback
            
        return total_logins / days_member

    # 4. Group by User and Apply the Custom Aggregator
    # We select the 'Time' column specifically to pass into our function
    frequency_series = login.groupby('Login Id')['Time'].agg(calculate_user_freq)
    
    return frequency_series
````

#### `cookies_p_value`

Complete the implementation of the function `cookies_p_value`, which takes in an integer `N` and returns the estimated p-value of your investigation upon simulating the null hypothesis `N` times. (The p-value is an estimate of the true theoretical p-value of your test since it relies on simulation.)

***Note***: Plot the null distribution and your observed statistic to check your work. (If you decide to plot, you may have to run `import matplotlib.pyplot as plt` or `import plotly.express as px`.)

````python
def cookies_null_hypothesis():
    """
    Returns reasonable choices for the null hypothesis.
    (This function was already loop-free).
    """
    return [1, 2]

def cookies_p_value(N):
    """
    Simulates the null hypothesis N times using Vectorization.
    """
    # 1. Define parameters
    n_cookies = 250
    prob_burnt = 0.04
    observed_burnt = 15
    
    # 2. Vectorized Simulation
    # Instead of looping N times, we pass size=N to generate an array of N outcomes at once.
    # simulated_stats will be an array like [8, 12, 10, 9, ...] with length N.
    simulated_stats = np.random.binomial(n=n_cookies, p=prob_burnt, size=N)
    
    # 3. Vectorized Comparison
    # We compare the entire array to the scalar 'observed_burnt' in one step.
    # This creates a boolean array (e.g., [False, False, True, ...])
    is_extreme = simulated_stats >= observed_burnt
    
    # 4. Calculate P-Value
    # In Python, np.mean() on a boolean array automatically treats True as 1 and False as 0.
    # This effectively calculates (Count of True) / (Total N).
    p_value = np.mean(is_extreme)
    
    return p_value
````

#### `bhbe_col`

To start, complete the implementation of the function `bhbe_col`, which takes in a DataFrame like `heroes` and returns a Boolean Series that contains `True` for characters that have **both** blond hair and blue eyes, and `False` for all other characters. 

***Note***: If a character's hair color contains the word `'blond'`, uppercase or lowercase, we consider their hair to be blond for the purposes of this question. Similarly, if a character's eye color contains the word `'blue'`, uppercase or lowercase, we consider their eye color to be blue for the purposes of this question.

Fix a significance level (i.e. p-value cutoff) of 1%.

#### `superheroes_observed_statistic`
Complete the implementation of the function `superheroes_observed_statistic`, which takes in a DataFrame like `heroes` and returns the observed test statistic.

#### `simulate_bhbe_null` 
Complete the implementation of the function `simulate_bhbe_null`, which takes in a DataFrame like `heroes` and a positive integer `N` and returns an array of length `N`, where each element is a simulated test statistic according to the null hypothesis.

***Hint***: Like in `superheroes_observed_statistic`, you'll need to use both `bhbe_col` and information in the `heroes` DataFrame to complete your simulation. Remember that you cannot use `for`-loops in this question.

<br>

#### `superheroes_p_value` 
Complete the implementation of the function `superheroes_p_value`, which takes in DataFrame like `heroes` and returns a list where:
* The first element is the p-value for the hypothesis test, using 100,000 simulations.
* The second element is `'Reject'` if you reject the null hypothesis and `'Fail to reject'` if you fail to reject the null hypothesis, at the 1% significance level.

````python
def superheroes_test_statistic():
    """
    Returns a list of integers corresponding to valid test statistics.
    The test expects this exact function name.
    """
    # 1: Difference in proportions (Directional/One-sided). Valid.
    # 2: Number of "good" BHBE characters (Equivalent to proportion). Valid.
    return [1, 2]

# 2. Helper Column Function
def bhbe_col(heroes):
    """
    Returns a Boolean Series: True if character is both blond and blue-eyed.
    """
    # case=False handles 'Blond' vs 'blond'. 
    # na=False handles missing values (NaN) safely.
    is_blond = heroes['Hair color'].str.contains('blond', case=False, na=False)
    is_blue = heroes['Eye color'].str.contains('blue', case=False, na=False)
    
    return is_blond & is_blue

# 3. Observed Statistic
def superheroes_observed_statistic(heroes):
    """
    Returns the observed proportion of 'good' characters among the BHBE subgroup.
    """
    # Filter for BHBE characters
    bhbe_mask = bhbe_col(heroes)
    bhbe_df = heroes[bhbe_mask]
    
    # Calculate proportion of 'good' alignment in this subgroup
    observed_prop = (bhbe_df['Alignment'] == 'good').mean()
    
    return observed_prop

# 4. Simulation (Vectorized - No loops)
def simulate_bhbe_null(heroes, N):
    """
    Simulates the null hypothesis N times.
    Returns an array of N simulated proportions.
    """
    # The Null assumes the BHBE group is just a random sample from the full population.
    # Population 'good' rate:
    prob_good_pop = (heroes['Alignment'] == 'good').mean()
    
    # Sample size (Size of the BHBE group):
    n_bhbe = bhbe_col(heroes).sum()
    
    # Vectorized Simulation:
    # Flip 'n_bhbe' coins, 'N' times, with P(Head) = prob_good_pop
    simulated_counts = np.random.binomial(n=n_bhbe, p=prob_good_pop, size=N)
    
    # Convert counts to proportions
    simulated_props = simulated_counts / n_bhbe
    
    return simulated_props

# 5. P-Value Calculation
def superheroes_p_value(heroes):
    """
    Returns [p_value, Decision String]
    """
    # 1. Get Observed
    obs_stat = superheroes_observed_statistic(heroes)
    
    # 2. Simulate 100,000 times
    sim_stats = simulate_bhbe_null(heroes, 100000)
    
    # 3. Calculate P-Value (Right-tailed test: "Greater than")
    p_value = np.mean(sim_stats >= obs_stat)
    
    # 4. Decision at 1% significance level
    if p_value < 0.01:
        decision = 'Reject'
    else:
        decision = 'Fail to reject'
        
    return [p_value, decision]
````

## Part 3: Permutation Testing

Recall, hypothesis tests answer questions of the form:

> I have a population distribution, and I have one sample. Does this sample look like it was drawn from the population?

While permutation tests answer questions of the form:

> I have two samples, but no information about any population distributions. Do these samples look like they were drawn from the same population?

### Question 6 – Orange Skittles 🟠

First, you will investigate if the machine that mixes together the Skittles of different colors might favor one color over another. Use a permutation test to assess whether, on average, bags made in Yorkville have the same number of orange skittles as bags made in Waco. Do this by implementing the functions described below.

<br>

#### `diff_of_means`

Complete the implementation of the function `diff_of_means`, which takes in a DataFrame like `skittles` and returns the **absolute difference** between the **mean** number of orange Skittles per bag from Yorkville and the **mean** number of orange Skittles per bag from Waco.

<br>

#### `simulate_null`

Complete the implementation of the function `simulate_null`, which takes in a DataFrame like `skittles` and returns one simulated instance of the test statistic under the null hypothesis. Note that this will involve shuffling the `'Factory'` column!

<br>

#### `color_p_value`

Complete the implementation of the function `color_p_value`, which takes in a DataFrame like `skittles` and calculates the p-value for the permutation test using 1000 trials.

<br>

Plot the observed statistic, along with the histogram for the simulated distribution, to check your work.

***Note***: In all functions, the default argument for `col` is `'orange'`. Your functions should still work for any color so that you can call it in later questions.

````python
def diff_of_means(skittles, col='orange'):
    """
    Calculates the absolute difference in mean count of 'col' between the two factories.
    """
    # Group by 'Factory' and calculate the mean for the specific color column
    # The factories are 'Yorkville' and 'Waco'
    means = skittles.groupby('Factory')[col].mean()
    
    # Calculate the absolute difference
    # means is a Series with index ['Waco', 'Yorkville'] (alphabetical)
    # diff = | Mean_A - Mean_B |
    diff = abs(means[0] - means[1])
    
    return diff

def simulate_null(skittles, col='orange'):
    """
    Returns one simulated test statistic under the null hypothesis.
    This involves shuffling the labels.
    """
    # 1. Shuffle the labels
    # We use sample(frac=1) to shuffle rows, and .values to strip the index.
    # If we didn't use .values, pandas would realign the index and undo the shuffle!
    shuffled_labels = skittles['Factory'].sample(frac=1, replace=False).values
    
    # 2. Assign shuffled labels to the data
    # We can create a temporary dataframe or just group the existing color data
    # by the new shuffled array.
    # This acts as if we randomly assigned each bag to a factory.
    
    # Group the actual color counts by the shuffled factory labels
    shuffled_means = skittles[col].groupby(shuffled_labels).mean()
    
    # 3. Calculate the test statistic for this shuffle
    simulated_diff = abs(shuffled_means[0] - shuffled_means[1])
    
    return simulated_diff

def color_p_value(skittles, col='orange'):
    """
    Calculates the p-value for the permutation test using 1000 trials.
    Also plots the histogram to check work.
    """
    # 1. Calculate the Observed Statistic (Real World)
    observed_diff = diff_of_means(skittles, col)
    
    # 2. Run the Simulation (1000 times)
    n_repetitions = 1000
    simulated_stats = []
    
    for i in range(n_repetitions):
        # Generate one fake statistic
        sim_stat = simulate_null(skittles, col)
        simulated_stats.append(sim_stat)
    
    # 3. Calculate P-Value
    # In a permutation test, the p-value is the proportion of shuffled differences
    # that are as large or larger than the observed difference.
    simulated_stats = np.array(simulated_stats)
    p_value = np.count_nonzero(simulated_stats >= observed_diff) / n_repetitions
    
    # --- Visualization (To check your work) ---
    plt.figure(figsize=(8, 5))
    plt.hist(simulated_stats, bins=20, density=True, alpha=0.6, color='skyblue', label='Null Distribution (Shuffled)')
    plt.axvline(observed_diff, color='red', linestyle='dashed', linewidth=2, label=f'Observed Diff ({observed_diff:.2f})')
    plt.title(f'Permutation Test for "{col}" Skittles')
    plt.xlabel('Absolute Difference in Means')
    plt.legend()
    plt.show()
    
    return p_value
````

**Are these distributions of colors similar?** Is the variation among the bags due to each factory making different amounts of each color?

Use a permutation test to assess whether the distribution of colors of Skittles made in Yorkville is statistically significantly different than those made in Waco. Set a significance level (i.e. p-value cutoff) of 0.01 and determine whether you can reject a null hypothesis that answers the question above using a permutation test with 1000 trials. For your test statistic, use the **total variation distance (TVD)**.

Refer to an example of a [permutation test](https://www.inferentialthinking.com/chapters/12/Comparing_Two_Samples.html) that uses the [TVD](https://inferentialthinking.com/chapters/11/2/Multiple_Categories.html) as the test statistic. Some guidance:

- Our previous permutation tests have compared the mean number of (say) orange Skittles in Yorkville bags to the mean number number of orange Skittles in Waco bags. The role of shuffling was to randomly assign bags to Yorkville and Waco.
- In this permutation test, we are **still** shuffling to randomly assign bags to Yorkville and Waco. The only difference is that after we randomly assign each bag to a factory, we will compute the **distribution** of colors among the two factories and find the TVD between those two distributions.

**Your job**: Complete the implementation of the function `same_color_distribution`, which takes in no arguments and outputs a hard-coded **tuple** with the p-value and whether you `'Reject'` or `'Fail to reject'` the null hypothesis.

````python
# 1. Helper to Calculate TVD
def calculate_tvd(df):
    """
    Computes the TVD between the two factories in the given DataFrame.
    Assumes df has 'Factory' and color columns.
    """
    # Sum up all colors for each factory (Get total counts)
    # Result is a DataFrame with 2 rows (Factories) and 5 columns (Colors)
    factory_counts = df.groupby('Factory')[['red', 'orange', 'yellow', 'green', 'purple']].sum()
    
    # Convert counts to Proportions (divide by row sum)
    # axis=0 is column, axis=1 is row. We divide each row by its sum.
    factory_props = factory_counts.div(factory_counts.sum(axis=1), axis=0)
    
    # Calculate TVD: 1/2 * Sum(|Prop_A - Prop_B|)
    # We take the difference between the two rows (Yorkville - Waco)
    diffs = factory_props.diff().iloc[-1].abs() # .diff() creates NaN in first row, so take last row
    tvd = 0.5 * diffs.sum()
    
    return tvd

# 2. Calculate Observed TVD
obs_tvd = calculate_tvd(skittles)
print(f"Observed TVD: {obs_tvd:.5f}")

# 3. Permutation Test (Shuffle Bags)
n_reps = 1000
simulated_tvds = []

# We create a copy so we don't mess up the original data
shuffled_df = skittles.copy()

for _ in range(n_reps):
    # Shuffle the Factory labels
    # We use .values so the index doesn't reset the order
    shuffled_labels = skittles['Factory'].sample(frac=1).values
    shuffled_df['Factory'] = shuffled_labels
    
    # Calculate TVD for this shuffled scenario
    sim_tvd = calculate_tvd(shuffled_df)
    simulated_tvds.append(sim_tvd)

# 4. Calculate P-Value
simulated_tvds = np.array(simulated_tvds)
p_value = np.count_nonzero(simulated_tvds >= obs_tvd) / n_reps

# 5. Determine Decision
decision = 'Reject' if p_value < 0.01 else 'Fail to reject'

print("-" * 30)
print(f"P-Value: {p_value}")
print(f"Decision: {decision}")
print("COPY THESE VALUES INTO YOUR FUNCTION")
print("-" * 30)
````
